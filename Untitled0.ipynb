```python
import pandas as pd
import numpy as np # Added import for numpy
url = "https://raw.githubusercontent.com/ManarOmar/New-York-Airbnb-2019/master/AB_NYC_2019.csv"
data = pd.read_csv(url)
data.head()
data.shape
data.info()
data["id"].is_unique
print(len(data))
print(data["host_id"].nunique())

columnas = ['price', 'minimum_nights', 'number_of_reviews','reviews_per_month','calculated_host_listings_count','availability_365']

data[columnas].describe()

numericas = data[columnas]
numericas = numericas.fillna(numericas.mean())

data.hist(bins = 20, figsize=(10,6))

plt.tight_layout()
plt.show()

data.hist(bins = 20, figsize=(10,6))
plt.tight_layout()
plt.show()

data['neighbourhood'].value_counts()
data['room_type'].value_counts()
data['neighbourhood_group'].value_counts()

media = data['reviews_per_month'].mean()

print('Media de reviews_per_month:', round(media,2))

data['reviews_per_month'] = data['reviews_per_month'].fillna(0)
data['reviews_per_month'].isnull().sum()

data['price_z'] = data['price'].fillna(0)
data['price_media']= data['price'].fillna(media)

data.boxplot(column='price', figsize=(8,10))
plt.ylabel('Precio por la noche (USD)')
plt.show()




# Fix the error: Calculate means only for numeric columns before filling NaNs
numeric_means = data.select_dtypes(include=np.number).mean()
data = data.fillna(numeric_means)
```

```python
precio_maximo = 1000
precio_minimo = 0
datos_limpios = data[(data['price'] > precio_minimo) & (data['price'] <= precio_maximo)].copy()
```

```python
datos_limpios=data[(data['price']>precio_minimo)&(data['price']<precio_maximo)].copy()
print('Antes:',len(data),'anuncios')
print('Después:',len(datos_limpios),'anuncios')
datos_limpios['price'].describe()
```

```python
corte_economico = 80
corte_medio = 200

def categoria_precio(price):
  if price <= corte_economico:
    return 'economico'
  elif price <= corte_medio:
    return 'medio'
  else:
    return 'premium'

datos_limpios['categoria']=datos_limpios['price'].apply(categoria_precio)
datos_limpios['categoria'].value_counts()

```

```python
#el histograma de precio ahora si se lee
datos_limpios['price'].hist(bins=50, figsize=(10,5))
plt.xlabel('Precio por noche (USD)')
plt.ylabel('Anuncios')
plt.title('Distribución de precios')
plt.show()
```

```python
datos_limpios.groupby('neighbourhood_group')['price'].agg(['count','mean','median'])
```
